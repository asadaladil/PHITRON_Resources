```python
# library import
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder,StandardScaler
```

```python
df=pd.read_csv("https://raw.githubusercontent.com/asadaladil/Others-Materials/refs/heads/main/insurance.csv")
print(df.shape)
print(df.columns)

target_variable='smoker'

df.head(10)
```

```python
print(df.isnull().sum())
df.describe()
```

```python
numerical_col=['age','bmi','children','charges']
categorical_col=['sex','smoker','region']

df_encoded=pd.get_dummies(df, columns=categorical_col, dtype=int,drop_first=True)
temp=df_encoded['smoker_yes']
df_encoded.drop('smoker_yes',axis=1,inplace=True)
df_encoded['smoker']=temp

df_encoded.head()
```

```python
X=df_encoded.drop(columns=[target_variable])
y=df_encoded[target_variable]

scaler = StandardScaler()

X_numpy=scaler.fit_transform(X)
y_numpy=y.to_numpy()

torch.manual_seed(100)
X_torch=torch.tensor(X_numpy,dtype=torch.float32)
y_torch=torch.tensor(y_numpy,dtype=torch.float32).view(-1,1)
X_train,X_test,y_train,y_test=train_test_split(X_torch,y_torch,test_size=0.2,random_state=42)
print(X_train.shape)
```

##Explaination:

- The Problem: Neural networks and mathematical models cannot perform calculations on NaN or Null values. So, we need to remove or fill the null value.

2. Encoding Categorical Variables
- Neural networks only understand numbers, not text like "Red", "Green", or "Blue". We use one-hot encoding to convert categories into numerical value.


3. Separating Features (X) and Target (y)
- In supervised learning, the model needs to learn the relationship between inputs and the desired output.


4. Splitting into Train and Test Sets
- If we train a model on all the data, it might just memorize the answers & it will cause overfitting. We won't know if it can actually generalize to new, unseen data.


5. Feature Scaling (StandardScaler)
- We scale all numerical features to have a mean of 0 and a standard deviation of 1. This puts all features on an equal playing field so the neural network optimizes smoothly.


6. Converting to NumPy Arrays
- Pandas DataFrames are great for data manipulation, but they carry a lot of overhead like indexes and column names.


7. Converting to PyTorch Tensors
- PyTorch cannot use NumPy arrays directly for training. It requires its own data structure called a Tensor, which is optimized to run massively parallel calculations on a GPU.

```python
class Model(nn.Module):
  def __init__(self,input_size,hidden_size,output_size):
    super().__init__()
    self.linear1=nn.Linear(input_size,hidden_size)
    self.relu=nn.ReLU()
    self.linear2=nn.Linear(hidden_size,output_size)
    self.sigmoid=nn.Sigmoid()

  def forward(self,X):
    out=self.linear1(X)
    out=self.relu(out)
    out=self.linear2(out)
    out=self.sigmoid(out)
    return out
```

```python
model=Model(X_torch.shape[1],32,1)
model(X_torch)
```

##Justification:
Number of Hidden Neurons:
- The number of hidden neurons controls the learning capacity of the model. If hidden_size is too small, the model suffers from underfitting and cannot learn complex patterns in the data. If it is too large, the model suffers from overfitting, where it memorizes the training data but fails to generalize to new, unseen data. It also makes training slower.


Choice of Activation Functions:
- I used two different activation functions here.

- ReLU (Hidden Layer): The Rectified Linear Unit is the industry standard for hidden layers. It introduces the necessary non-linearity so the network can learn complex, curved boundaries rather than just straight lines. It is also computationally cheap and prevents the vanishing gradient problem, allowing the model to learn quickly during backpropagation.

- Sigmoid (Output Layer): The Sigmoid function mathematically squashes whatever raw number the network outputs into a strict range between 0.0 and 1.0. This allows the model to interpret the final output as a probability.

Code to Print Trainable Parameters:
- To print the exact number of parameters, we have to instantiate the model with real numbers. Here is how you calculate and print the total parameters that PyTorch will be tweaking during training:

- Where that number comes from (8,32,1):

- linear1: (8 inputs * 32 neurons) + 32 biases = 256 parameters

- linear2: (32 inputs * 1 neuron) + 1 bias = 33 parameters

- Total: 256 + 33 = 289 parameters.

## Write Answer 03:

- Loss function: Binary Cross Entropy(BCE)
- Optimizer: Stochastic Gradient Descent (SGD):

Justification (Regression vs. Classification):

- The model ends with a nn.Sigmoid() layer, which results the output into a probability between 0.0 and 1.0. Because the model is predicting discrete categories rather than a continuous number, this is a binary classification task. BCE is the mathematically optimal loss function for this. It calculates the logarithmic distance between your model's predicted probability and the actual true label (0 or 1).

- If this were a regression task like predicting a house's price in dollars, I would use Mean Squared Error (nn.MSELoss). Using MSE for classification is a poor choice because it doesn't heavily penalize the model for being entirely wrong but highly confident. BCE applies a massive logarithmic penalty when the model predicts 0.99 for a label that is actually 0, forcing the network to learn rapidly from its worst mistakes.

Justification (Nature of the dataset):

- Standard datasets often contain noise or outliers. SGD updates the model's weights using only a small subset of the data at a time, it introduces a bit of noise into the training path. This noise actually acts as a good natural regularizer, preventing the model from memorizing the training data and helping it generalize better to unseen testing data.


```python
loss_function=nn.BCELoss()
optimizer=torch.optim.SGD(model.parameters(), lr=0.01)
```

## Write Answer 04:

```python
epochs=100
loss_history=[]

for epoch in range(epochs):
    y_pred=model.forward(X_train)
    #print(y_pred.shape,y_train.shape)
    loss=loss_function(y_pred,y_train)
    loss.backward()
    optimizer.step()
    loss_history.append(loss.item())
    if (epoch+1)%10==0:
        print(f'Epoch [{epoch+1}/{epochs}], Loss: {loss.item():.4f}')

    optimizer.zero_grad()

print("Training complete!")
```

##Pipeline:

Forward Pass:
- Input input data is pushed through the network. It gets multiplied by the weights in linear1, passes through the ReLU activation, goes through linear2, and finally gets squashed by the Sigmoid activation. The output is the model's current "best guess" (a probability between 0.0 and 1.0).

Loss Computation
- The BCELoss function takes the model's predictions and compares them directly to the true labels. It outputs a single scalar value representing how "wrong" the model was on this batch of data.

Backward Pass
- PyTorch's autograd engine performs backpropagation. It uses calculus to travel backward through the network, calculating the gradient for every single weight and bias. These gradients tell the optimizer exactly which direction to adjust each parameter to lower the error.

Parameter Update
- The SGD optimizer steps in, takes the gradients calculated in the previous step, and applies them to the network's weights. It updates the parameters by taking a small step down the gradient slope.

Gradient Reset
- As PyTorch accumulates gradients, it means if we don't explicitly clear the gradients from the previous epoch, the new gradients will just be added on top of the old ones. This will completely ruin the math and prevent the model from learning. We call this first to ensure we are starting with a clean slate.

## Write Answer 07:

Gradients be reset every epoch:
- PyTorch accumulates gradients. When calling loss.backward(), it doesn't replace the old gradients; it adds the newly calculated gradients.

If the learning rate is too high:
- The learning rate determines the step size for the optimizer takes down or the slope of the loss function.
- If the step is too high, the optimizer will step completely over the lowest point of the error valley and land higher up on the opposite side.
- It will diverge from one side of another side.


If the learning rate is too small:
- The optimizer takes little steps. The loss will go down, but it will take an eternity. For standard learning rate it would take 100 epochs might now take 50,000 epochs for slow learning rate.
- Small learning rate does not have proper power to reach out the local or global minimum point of error.

Define layers inside the constructor (__init__) and not inside forward():
- __init__ is for creating the object of a class.
- When defining nn.Linear() inside the constructor, PyTorch allocates memory for the weights and biases of that layer exactly once. These parameters stay in memory.

- forward() is just the roadmap.
- This function executes every single time passing a batch of data into the model. If I define nn.Linear() inside forward(), I am telling PyTorch to completely modify the old layer and create a new one with randomized, untrained weights every single time a new batch of data arrives.
