# Question 04: [ Marks 15 ]

Write the perceptron weight update rule and explain it intuitively.
Why are the weights updated only when the prediction is incorrect?

## Write Answer 04:
- The perceptron weight update formula is w_new=w_old + eta*(y-y_hat)*x_i
- Where:
- w_new: The weight connecting input x_i to the neuron
- eta: The Learning Rate that controls step size.
- y: The actual correct label (1 or 0).
- y_hat: The perceptron's predicted output (1 or 0).
- x_i: The input value.

- error= y-y_hat.
- if the error is (1-0)=+1, the prediction is wrong. So, we need to update the old weight. It increases the weight and making the sum larger next time so it is more nearly to output 1.
- if the error is (0-1)=-1, the also prediction is wrong. So, we still need to update the old weight. It decreases the weight, making the sum smaller next time so it is less nearly to fire or nearly to output 0.

- For 0 error, weights are not updated. w_new=w_old

-Indeed,  If the perceptron correctly classifies a data point, the current decision boundary line is already in the right place for that point. Move it when wrong prediction by providing stability that previously got correct prediction.

# Question 05: [ Marks 10 ]

Why does a single-layer perceptron fail to solve the XOR problem?
What does this limitation tell us about the need for multilayer neural networks?

## Write Answer 05:
- A single-layer perceptron is mathematically limited to a Linear Classifier. Its decision boundary is strictly a straight line (in 2D) or a flat plane (in 3D). It separates data based on the simple equation w₁x₁ + w₂x₂ + b = 0.
- The XOR Data: plotting the XOR truth table on a graph:
- Class 0 (False): Points are at (0,0) and (1,1).
- Class 1 (True): Points are at (0,1) and (1,0).
- If we draw a line to separate (0,0), we  cut off one of the True points.
- If we draw a line to separate (1,1), we cut off the other True point.
- It is geometrically impossible. We cannot separate these points with one straight line. The XOR dataset is not linearly separable.

- This failure is what led to the development of Multilayer Perceptrons and modern Deep Learning as real-world data  is rarely simple enough to be split by a single straight line.

# Question 06: [ Marks 20 ]

What is meant by the “perceptron Learning Rule” or weight adjustment method?
Why can adjusting one weight affect the classification of other points?

## Write Answer 06:

- The Perceptron Learning Rule is the algorithm used to automatically teach a perceptron by iteratively adjusting its weights based on its errors.
- The rules are
- Give the perceptron an input x.
- Compare its predicted output y_hat with the actual target y.If the prediction is wrong: Update the weights w and bias b to reduce the error.If the prediction is correct: Do nothing.
- The formula is:
 w_new=w_old + eta*(y-y_hat)*x_i
- here eta is learning rate.
- The weights w control the angle or slop of the line and the bias b controls the position of the line.
- If I push down on one end of the line to make it fit a specific point, the line line moves and consequently the other end of the line goes up. It may shift or deviate the line away from another point that was previously correct.

- As the Perceptron uses a single straight line to classify the dataset, we cannot rotate or shift just one part of the line to fit a specific point. Any change we make to the line's angle affects its position everywhere. As a result, it breaks the classification for other points. This is why we use a small Learning Rate—to make small changes instead of big learning rate. Small learning rate helps to converge the line into minimum error point.

# Question 07: [ Marks 20 ]

What are the main limitations of a single-layer perceptron?
How does the idea of using more than one layer (MLP) help overcome these limitations at a high level accoriding to you?



## Write Answer 07:

- The main limitations of a single layer perceptron are:
1. It is strictly linear classifier.
2. It can only draw a single line on a flat sheet.
3. Most of the real-world data is non-linear. So, only single layer perceptron cannot solve all the problems.

- A Multi-Layer Perceptron (MLP) introduces hidden layers between the input and output.
- If we consider a 2 layer perceptron on an example:
1. Neuron 1 draws a line on the left of the sheet.
2. Neuron 2 draws a line on the right of the sheet.
3. If we combine them then two lines will form a "V" shape or a region. By adding enough neurons, we can build any shape.

- In MLP, we use non-linear activation function. With non-linearity it is easy to bend the system. The first layer draws lines, the activation function bends them, and the next layer combines them. As a result, the decision boundary can now curve, loop, and twist to fit perfectly around complex data dataset.

- It is easy to perform feature engineering using deep learning. If we consider a handwritten digit detection task, it will look like:
1. Layer-1: Raw data.
2. Layer-2: Combines pixels to find edges.
3. Layer 3: Combines edges to find shapes.
4. Output Layer: Combines shapes to recognize objects, in this case digits.

- So, the idea of using MLP overcomes the linearity problem to solve the complex and complicated dataset.
