# Question 01: [ Marks 10 ]
Explain gradient descent in detail. What is the main idea behind it? How does it help in minimizing the loss function? Explain the role of learning rate and describe what happens if the learning rate is too high or too low.

**Write** Answer 01:

**Gradient Descent**
- Gradient Descent is an optimation algorithm for deep learning.
- It helps to tuning the slop and bias for linear relation function by iteratively moving in the direction of local minimum point.
- In general, in machine learning, it optimezes and update the parameters of a model to reduce the error or loss.

**Main Idea**

The gradient is simply a vector of partial derivatives. In calculus, the derivative of a function determines the slope of the function at the specific point.

- Positive Gradient: The function is increasing.
- Negative Gradient: The function is decreasing.
- Zero Gradient: it is at a local minimum point.

Since the algorithm minimizes the loss, it goes in the opposite direction of the gradient.

**Minimizing loss function**

- Gradient descent starts with random initial values for the model's parameters. Then these steps are followed:

1. Compute the loss using the current values of the parameters.

2. Determine the direction and magnitude of the steepest descent by calculating the gradient of the loss function with respect to each parameter.

3. Adjust the parameters in the direction opposite to the gradient to reduce the loss. Learning rate helps to adjust the changing direction.

4. Continue the process until the loss converges to a minimum value or old value and new become same.

**Role of Learning Rate**

- Learning rate is a hyperparameter that determines the size of the steps.
- It is an important part for neural network.
1. If the Learning Rate is Too High, the steps are too large, it might overshoot the actual point.
2. If the Learning Rate is Too Low, the steps are too small, the algorithm will reach the minimum, but it will take a very long time to converge. This wastes computational resources and time.

## Write Answer 04:

**Sigmoid Activation Function**

- It is a mathematical function that takes any real-valued number and maps it into a range between 0 and 1.
- The formula is:
$$\sigma(x) = \frac{1}{1 + e^{-x}}$$

**Usuage**

- It is used in binary classification problem.
- As the output is limited between 0 and 1, it is perfect for predicting probabilities between 2 classes.

**Advantages & Disadvatanges**

- Advantage:
1. Output Range between [0,1].
2. The function is differentiable everywhere, resulting in a smooth S-shaped curve that prevents jumps in output values.
- Disadvantage:
1. It is computationally expensive as it is an exponential function.
2. the output is always positive. no negative output exists.

# Question 05: [ Marks 10 ]
Explain the softmax activation function in simple words. Why is it used in multi-class classification problems? How is it different from sigmoid?

## Write Answer 05:

**Softmax Activation Function**

- Softmax is an activation function that turns a list of raw numbers into a list of probabilities that sum up to exactly 1 or 100%.
- It is used in multiclass classification problem where each neuron tries to predict the class having the largest probability value.

**Usage**

- Softmax is standard activation function function for multi-class problem.
- There is a property , called mutual exclusivity that helps to compete among the classess. If there is 3 class like 'dog','cat','bird', the probability of 'dog' goes up, the probability of 'cat and 'bird' must go down. This reflects the reality that the image can't be both 100% Cat and 100% Dog. It will always fall into a specific class.
- The Formula is: $$\sigma(z)_i = \frac{e^{z_i}}{\sum_{j=1}^{K} e^{z_j}}$$

**Different from Sigmoid**

1. Sigmoid has independent output relation and the softmax has interdependent output relationship.
2. In sigmoid, probabilities do not affect each other. In softmax, probabilities affect each other.
3. In sigmoid, the sum of probababilities can be anything. In softmax, the sum of probabilities is exactly 1.
4. Sigmoid is useful for binary classification problem. Softmax is useful for multi-class classification problem.
5. The sigmoid formula: $$\sigma(x) = \frac{1}{1 + e^{-x}}$$ The softmax formula: $$\sigma(z)_i = \frac{e^{z_i}}{\sum_{j=1}^{K} e^{z_j}}$$


# Question 06: [ Marks 20 ]

Explain the structure of an Artificial Neural Network in detail. Describe the role of the input layer, hidden layers, and output layer. How do weights, biases, and activation functions work together to transform input data into predictions? Illustrate your explanation with a simple example.

## Write Answer 06:

**Structure of an ANN**

- An Artificial Neural Network (ANN) is a computational model which inherited by the biological neural networks in the human brain.
- It consists of interconnected units called neurons.
- It is organized into layers.
- These networks are designed to recognize patterns, whether they are visual, sequential, or numerical.

20260218_132101.jpg

**The role of Layers**

1. Input Layer:
- Role: This is the first phase of the network. It receives the raw data from the outside world.
- Function: No computation happens here. The input layer simply passes the numbers to the next layer.
- Size: The number of neurons equals the number of features in the data.

2. Hidden Layer:
- Role: These layers are middle point between the input and output layers. This is where the learning and computation happens.

- Function: It performs mathematical transformations on the input to extract features.

3. Output Layer:
- Role: This layer produces the final prediction or result.
- Size: The number of neurons depends on the problem. If it is Regression, 1 neuron for predicting value. If it is binary classification, 1 neuron for yes or no class. If it is multi-class classification, 1 neuron per class.

**Transform input data into predictions**

 1. Weights ($w$)
 - Every connection between neurons has a weight. This value determines how much effect one neuron has on another.
 - High Weight: The input is very important.
 - Low Weight: The input is ignored.
 - Negative Weight: The input reduces the output.

 2. Biases ($b$)
- Bias is an extra value added to the weighted sum.
- It allows the neuron to shift its activation function left or right.

3. Activation Function ($f$)
- It is called the decision maker.
- After the weights and biases are applied, the result is a simple number. The activation function decides whether this neuron should fire or not.
- It introduces non-linearity, allowing the network to learn complex, curved boundaries instead of just straight lines.

**Example**

Predict the price of a house:

Input Layer:
- Area
- Number of Bedrooms

Hidden Layer:
- Neuron A might learn to look for family size by combining Area + Bedrooms.- - - Neuron B might learn to look for luxury getting high Area but few Bedrooms.

Output Layer:
- Predicted Price.
