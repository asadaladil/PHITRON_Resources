- Backpropagation is an algorithm which is used to calculate the gradient of the loss function with respect to each weight in the neural network.
- It helps to update the weights and minimize the error.
- It starts from output layer, then back to the hidden layer and finally moves to the input layer for being weight updated correctly.
- So, after completing forward pass, backpropagation helps to build up a good neural network for a problem.

# How error is propagated from output layer to the hidden layer:

1. Through forward pass the network calculates the output.
2. It makes a prediction there.
3. then calculate the loss difference between prediction and target by squaring the difference.
4. Then we have to calculate the partial derivative of the loss function with respect to all the weights and biases of the different hidden layer.
5. After performing partial derivative put the value in weight and bias updation formula:
W_new=W_old-eta*(del_L/del_W_old), b_new=b_old-eta*(del_L/del_b_old)
6. This approach is continued until it reaches the input and get the better result for the model.

## Discuss the vanishing gradient problems in deep neural networks. Explain why they occur during backpropagation, how activation functions contribute to this issue, and suggest at least three techniques to overcome these problems.


## Write Answer 03:
- The vanishing gradient problem occurs when the value of partial derivatives become extremely small as they are backpropagated from the output layer toward the input layer.

Reason:
- The issue lies in the chain rule.
- If partial derivatives are small, multiplying them repeatedly causes the total product to decrease exponentially. As a result W_new=W_old, no difference between new and old weight.
- The problem arises if there is too many layers.
- Using Activation function like sigmoid, tanh activation.

Overcome technique:
- We must reduce complexity.
- By using ReLU activation function.
- Using Batch Normalization.

# Roles of the Core Components:
Neurons:

# Architectural Comparison:
Artificial Neural Network (ANN):
- Structure: Consists of an input layer, one or more hidden layers, and an output layer.
- Data Handling: ANNs typically expect 1D vectors. To process an image in an ANN, we must flatten it to 1D (Ex- a $28 \times 28$ image becomes a single $784$-element vector).
- Parameters: The number of weights grows exponentially with input size, making it computationally expensive for high-resolution data.

Convolutional Neural Network(CNN):

- Structure: Introduces specialized layers named Convolutional layers to extract features, Pooling layers to reduce dimensionality, and Fully Connected layers at the end for classification.
- Parameter Sharing: Instead of having a unique weight for every pixel, CNNs use small filters that slide across the image. The same filter is used across the entire input, reducing the number of parameters.

# Working Principles:
ANN:
- Feature Learning: Learns global pattern.
- Spatial Invariance: If an object moves to a different corner of an image, an ANN usually fails to recognize it unless trained on that specific position.
- Transformation: Matrix multiplication followed by a non-linear activation.

CNN:
- Feature Learning: Learns local pattern
- Spatial Invariance: The filters slide across the whole image, a CNN can recognize a feature regardless of where it appears.
- Transformation: Convolution or mathematical sliding window followed by downsampling or Pooling.

# Advantages and Limitations
ANN:
- Advantages: Excellent for structured, tabular data like CSV files or databases. It is Relatively simple to implement for basic regression and classification tasks.
- Limitations: It suffers from the Curse of Dimensionality with images; loses spatial information; fall to vanishing gradients in very deep fully-connected structures.

CNN:
- Advantages: Extreme efficiency in feature extraction. It requires fewer parameters than a flattened ANN for the same image size.It is like state-of-the-art performance in computer vision.

- Limitations: It requires massive amounts of labeled data to train from scratch which computationally intensive to train and requires GPUs. It performs poorly on tabular data where feature order doesn't matter.

# Typical Use Cases:
Tabular Data & Pattern Recognition (ANN):
- Financial Forecasting: Predicting stock prices or medical costs based on numerical features.

- Classification: Simple digit recognition when the images are small and centered.

Image Processing & Computer Vision (CNN)
- Object Detection: Identifying multiple objects within a single frame.
- Medical Imaging: Detecting tumors in X-rays or MRIs.
- Facial Recognition: Identifying specific individuals in complex environments.

## Explain different activation functions used in neural networks, including Sigmoid, Tanh, and ReLU. Provide their mathematical formulas, advantages, disadvantages, and suitable use cases. Also discuss why ReLU is widely used in deep learning.

## Write Answer 06:

Activation Function:
- It determines whether a neuron should be activated or not.
- The primary purpose is to introduce non-linearity into the network, which allows it to learn complex relation between inputs and outputs.

Sigmoid Activation Function:

- It determines input values into a range between 0 and 1.
- Formula: $\sigma(x) = \frac{1}{1 + e^{-x}}$
- Advantages: Smooth gradient, output is probability-based which is ideal for binary classification.
- Disadvantages: Vanishing Gradient- For very high or low inputs, the gradient is almost zero, which stops the network from learning.
Computationally Expensive- Uses the exponential function.
Not Zero-Centered: The output is always positive.
- Use Case: It is mostly used in the output layer of binary classification models.

2. Tanh:
- It determines values into a range between -1 and 1.
- Formula: $\tanh(x) = \frac{e^x - e^{-x}}{e^x + e^{-x}}$
- Advantages: Zero-Centered- The mean of the output is closer to zero, which helps in centering the data for the next layer and makes optimization easier.Stronger gradients than Sigmoid.
- Disadvantages: It still suffers from the Vanishing Gradient problem when the neuron values are far from 0.
- Use Case: Often used in hidden layers of Recurrent Neural Networks.

3. ReLU:
- It returns 0 if the input is negative and the input itself if it is positive.
- Formula: $f(x) = \max(0, x)$
- Advantages: Computationally Efficient- It only requires a simple thresholding x>0, which is much faster than calculating exponentials.Reduces Vanishing Gradient- For x>0, the gradient is always 1, allowing gradients to flow freely through deep layers.Sparsity- It outputs 0 for negative values, meaning only a subset of neurons are active at any time.
- Disadvantages: Dying ReLU Problem- If a neuron's weight update pushes it into the negative range for all inputs, it will always output 0. The gradient becomes 0, and that neuron will dead.
- Use Case: The default choice for hidden layers in almost all modern deep neural networks like CNNs, MLPs.

ReLU is widely used in Deep Learning:

1. Symmetry with Gradient Descent: Because the gradient of ReLU is 1 for all positive values, it prevents the gradient from shrinking exponentially as it moves through 50 or 100 layers. This was the key breakthrough that allowed for deep learning.

2. Hardware Friendliness: Unlike Sigmoid or Tanh, which require complex transcendental math , ReLU can be implemented in hardware with a simple comparator and a multiplexer. This makes it much easier to implement on an FPGA or custom VLSI chip.

3. Faster Convergence: The networks using ReLU converge up to 6x faster than those using Tanh because they don't saturate in the positive direction.

## Explain the concept of pooling in CNNs. Compare MaxPooling and AveragePooling in terms of operation, advantages, and impact on feature extraction. Discuss how pooling helps in reducing overfitting and computational complexity.

# Write Answer 08:

Concept of pooling in CNNs:
- Pooling also known as downsampling is a non-linear process that reduces the spatial dimensions of an input volume.
- It involves sliding a fixed-size window across the feature map with a specific stride.
- It performs a summary operation on the values within that window.

MaxPooling:
- Operation: It selects the maximum value from the region covered by the filter.
- Impact on Features: It acts as a strongest feature extractor. If a specific pattern is detected anywhere in the window, Max Pooling preserves that high value.
- Advantage: It is very effective at discarding noise and focusing on the most prominent features of the image.

AveragePooling:
- Operation: It calculates the average value of all pixels in the window.
- Impact on Features: It acts as a smoothing" filter. Instead of picking the highest signal, it considers all signals equally.
- Advantage: It preserves the global structure of the image better than Max Pooling and it is often used in the final layers of a network to prepare data for classification.

Reducing Computational Complexity:

- By reducing the spatial size of the feature maps, pooling drastically reduces the number of parameters and the amount of computation required in subsequent layers.
- Example: A $2 \times 2$ pooling layer with a stride of 2 reduces the number of pixels by 75%. This makes it much easier to implement on hardware like the FPGAs and as it minimizes memory bandwidth requirements.

Reducing Overfitting:
- Pooling introduces Spatial Invariance. Because it summarizes a region into a single value, the network becomes less sensitive to the exact location of a feature.
- If an object shifts by a few pixels, the Max Pooled output will likely remain the same.By providing this significant representation, the model learns the general concept of the object rather than memorizing exact pixel coordinates.

## Design and implement a deep Convolutional Neural Network (CNN) for a multi-class image classification problem using PyTorch. The network should consist of four convolutional blocks, where each block includes two convolution layers followed by activation and normalization, and a pooling layer at the end. The number of filters should increase progressively across blocks. After the convolutional part, flatten the feature maps and connect them to fully connected layers with dropout for regularization, and produce final outputs corresponding to K classes. Also explain the role of each part of the network in detail.

# Write Answer 10:

```python
import torch
import torch.nn as nn

class MyCNN(nn.Module):
    def __init__(self, num_classes):
        super().__init__()

        self.block1 = self._new_block(3, 32)
        self.block2 = self._new_block(32, 64)
        self.block3 = self._new_block(64, 128)
        self.block4 = self._new_block(128, 256)

        self.classifier = nn.Sequential(
            nn.Flatten(),
            nn.Linear(256 * 8 * 8, 512),
            nn.ReLU(),
            nn.Dropout(0.4),
            nn.Linear(512, num_classes)
        )

    def _new_block(self, in_channel, out_channel):
        feature = nn.Sequential(
            nn.Conv2d(in_channel, out_channel, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.BatchNorm2d(out_channel),
            nn.Conv2d(out_channel, out_channel, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.BatchNorm2d(out_channel),
            nn.MaxPool2d(kernel_size=2, stride=2),
        )
        return feature

    def forward(self, x):
        x = self.block1(x)
        x = self.block2(x)
        x = self.block3(x)
        x = self.block4(x)
        x = self.classifier(x)
        return x

model = MyCNN(num_classes=10)
print(model)
```

# Role of Each Component
1. The Convolutional Blocks:
- Each block serves as a hierarchical feature extractor.
- Dual Convolution Layers: Using two consecutive $3 \times 3$ convolutions allows the network to develop a more complex, non-linear representation of features compared to a single larger filter.
- Progressive Filters: By increasing the filters ($32 \to 64 \to 128 \to 256$) because as the spatial resolution decreases via pooling, we want to increase the "depth" to capture a larger variety of complex patterns.

2. Normalization and Activation:
- Batch Normalization- This layer normalizes the output of the convolutions. It stabilizes the learning process, allows for higher learning rates, and acts as a mild form of regularization.
- ReLU Activation: This introduces the necessary non-linearity, enabling the network to learn relationships beyond simple linear transformations.

3. Pooling Layer
- Max Pooling: At the end of each block, we use a $2 \times 2$ MaxPool layer. This reduces the spatial dimensions by half, which controls computational costs and provides translational invariance.
4. The Classifier
- Flattening: This converts the 3D feature maps into a 1D vector so they can be processed by standard dense layers.
- Fully Connected Layers: These layers act as the reasoning part of the brain. They take the high-level features extracted by the convolutional blocks and combine them to determine which class the image belongs to.
- Dropout: During training, Dropout randomly turns off 40% of the neurons. This prevents the model from relying too heavily on any single neuron, forcing it to learn more robust features and significantly reducing overfitting.
